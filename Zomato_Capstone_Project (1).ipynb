```python
df.drop(columns=["zipcode","url","currency","takeaway","photo_count","res_id",'address'],axis=1,inplace = True )
```

```python
lower = df['timings'].str.lower()

condition_pm = lower.str.contains('pm', na=False) & lower.str.contains('am', na=False)
condition_am = lower.str.contains('am', na=False) & lower.str.contains('pm', na=False)

df['timings'] = 'Morning to Night'
df.loc[condition_am, 'timings'] = 'Morning till Noon'
df.loc[condition_pm, 'timings'] = 'Noon to Night'
```

```python
df["country_id"] = "India"
```

```python
df['cuisines'] = df['cuisines'].str.split(',').str[0]
```

```python
df['highlights'] = df['highlights'].str.split(',').str[1]
```

```python
df["rating_text"].unique()
```

```python
df['rating_text'] = df['rating_text'].replace({
    'Very Good': 'Very Good', 'Excellent': 'Excellent', 'Good': 'Good', 'Average': 'Average',
    'Not rated': 'Not rated', 'Poor': 'Poor',
    'Dobrē': 'Good', 'Wybitnie': 'Excellent','Excelente':"Excellent",'Eccellente':"Excellent",'Sangat Baik': 'Excellent', 'Ottimo': 'Excellent','Bardzo dobrze': 'Excellent','Muito Bom': 'Very Good', 'Velmi dobré': 'Very Good','Baik': 'Very Good',
    'Skvělá volba': 'Excellent', 'Bom': 'Good', 'İyi': 'Good','Bueno': 'Good', 'Çok iyi': 'Very Good','Priemer': 'Good','Biasa': 'Good','Buono': 'Very Good',"Veľmi dobré":"Very Good",
    'Harika':'Excellent', 'Terbaik': 'Excellent', 'Skvělé': 'Excellent', 'Průměr': 'Average',
    'Muito bom': 'Very Good', 'Dobrze': 'Good',"Dobré":"Good",'Média': 'Average','Media': 'Average',
    'Velmí dobré': 'Very Good', 'Vynikajúce': 'Excellent', 'Średnio': 'Average', "nan":'Not rated',"Nan":"Not Rated",'Scarso': 'Poor','Promedio': 'Average','Muy Bueno': 'Very Good','Ortalama': 'Good'})
```

```python
df["rating_text"].unique()
```

```python
df['cuisines'] = df['cuisines'].fillna('Unknown')
df['highlights'] = df['highlights'].fillna('Unknown')
```

```python
plt.figure(figsize=(35,5))
sns.barplot(data=df, x="establishment", y="aggregate_rating",palette="Set2")
plt.show()
```

```python
city_counts = df['city'].value_counts().head(50)
top_cities = city_counts.index.tolist()
df_top = df[df['city'].isin(top_cities)]
avg_ratings = df_top.groupby("city")["aggregate_rating"].mean().reindex(top_cities).reset_index()
plt.figure(figsize=(17, 6))
plt.bar(avg_ratings["city"], avg_ratings["aggregate_rating"], color="#33a02c")
plt.xlabel("City")
plt.ylabel("Average Rating")
plt.title("Average Restaurant Ratings by City (Top 20 Cities)")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
fig = px.scatter_mapbox(
    df,
    lat="latitude",
    lon="longitude",
    hover_name="name",
    hover_data=["city"],
    color="city",
    zoom=4,
    height=700,
    mapbox_style="carto-positron")
fig.update_layout(title='Restaurants in India (each dot = one restaurant)')
fig.show()
```

```python
df_city = df.groupby(['city', 'latitude', 'longitude']).size().reset_index(name='restaurant_count')
fig = px.scatter_mapbox(df_city,lat="latitude",lon="longitude",size="restaurant_count",color="restaurant_count",hover_name="city",hover_data=["restaurant_count"],zoom=4,height=700,mapbox_style="carto-positron")
fig.update_layout(title='Number of Restaurants per City in India')
fig.show()
```

```python
df_city_counts = df.groupby('city')['name'].count().reset_index(name='restaurant_count')
df_city_counts = df_city_counts.sort_values('restaurant_count', ascending=False)
plt.figure(figsize=(22, 7))
plt.bar(df_city_counts['city'], df_city_counts['restaurant_count'], color='skyblue')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.title('Number of Restaurants in Each City')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
cuisine50 = df['cuisines'].value_counts().head(25)
top_cuisine = cuisine50.index.tolist()
df_top2 = df[df['cuisines'].isin(top_cuisine)]
df_city_counts = df_top2.groupby('cuisines')['name'].count().reset_index(name='restaurant_count')
df_city_counts = df_city_counts.sort_values('restaurant_count', ascending=True)
plt.figure(figsize=(20,10))
plt.barh(df_city_counts['cuisines'], df_city_counts['restaurant_count'], color='purple')
plt.xlabel('Cuisine')
plt.ylabel('Number of Restaurants')
plt.title('Most Popular Cuisines offered by Restaurants')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18, 8))
sns.stripplot(data=df_top2, x="cuisines", y="aggregate_rating", order=df_top2['cuisines'].value_counts().index, jitter=True, alpha=0.5)
plt.xticks(rotation=45, ha='right')
plt.title("Aggregate Rating per Cuisine (Strip Plot)")
plt.show()
```

```python
plt.figure(figsize=(18, 8))
sns.boxplot(data=df_top2, x="cuisines", y="aggregate_rating", order=df_top2['cuisines'].value_counts().index)
plt.xticks(rotation=45, ha='right')
plt.title("Aggregate Rating Distribution by Cuisine")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.stripplot(data=df, x="price_range", y="aggregate_rating", alpha=0.5, jitter=True)
plt.xlabel("Price Range")
plt.ylabel("Aggregate Rating")
plt.title("Aggregate Rating vs Price Range (Strip Plot)")
plt.tight_layout()
plt.show()
```

```python
avg_cost = df.groupby('price_range')['average_cost_for_two'].mean().reset_index()
plt.figure(figsize=(8, 5))
plt.stem(avg_cost['price_range'], avg_cost['average_cost_for_two'])
plt.xlabel('Price Category')
plt.ylabel('Average Cost for Two')
plt.title('Lollipop Chart: Avg Cost for Two')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='price_range', y='average_cost_for_two', palette='Pastel1')
plt.xlabel('Price Range')
plt.ylabel('Average Cost for Two')
plt.title('Cost Distribution by Category (Violin Plot)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11,5))
df_counts = df["name"].value_counts().reset_index().head(20)
df_counts.columns = ['name', 'count']
plt.bar(df_counts['name'], df_counts['count'])
plt.xticks(rotation=45, ha='right')
plt.xlabel("Name of Restaurants")
plt.ylabel("Number of Outlets")
plt.title("Number of Restaurants")
plt.show()
```

```python
plt.figure(figsize=(10,8))
wedges, texts, autotexts = plt.pie(
    df_counts['count'],
    labels=df_counts['name'],
    autopct='%1.1f%%',
    startangle=190,
    wedgeprops=dict(width=0.5)
)
plt.title("Top 20 Restaurants (Number of Outlets) — Donut Chart")
```

```python
df_counts = df['name'].value_counts(ascending=False).head(20)
top_restaurant_names = df_counts.index.tolist()
avg_ratings = df[df['name'].isin(top_restaurant_names)].groupby('name')['aggregate_rating'].mean()
avg_ratings = avg_ratings[top_restaurant_names]

plt.figure(figsize=(11,5))
plt.bar(top_restaurant_names, avg_ratings)
plt.xticks(rotation=45, ha='right')
plt.xlabel('Name of Restaurants')
plt.ylabel('Aggregate Rating')
plt.title('Ratings of Top 20 Restaurants with Highest Number of Outlets')
plt.tight_layout()
plt.show()
```

```python
df_avg = df.groupby("highlights")["aggregate_rating"].mean().reset_index()

df_avg = df_avg.sort_values(by="aggregate_rating", ascending=False)

plt.figure(figsize=(17,6))
sns.lineplot(x="highlights", y="aggregate_rating", data=df_avg, marker="o")
plt.xticks(rotation=45, ha="right")
plt.xlabel("Highlights")
plt.ylabel("Average Rating")
plt.title("Correlation Between Highlights and Average Rating")
plt.grid(True)
plt.show()
```

```python
from wordcloud import WordCloud
categories = ["Excellent", "Very Good", "Good", "Average", "Poor", "Not Rated"]
review_counts = df['rating_text'].value_counts()
review_counts = review_counts.reindex(categories, fill_value=0)

word_freq = dict(review_counts)
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud of Review Categories")
plt.show()
```
